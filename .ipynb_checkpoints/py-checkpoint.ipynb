##### LIST

```python
insan=["meyra", "maden", 21]
print ( insan)
print( insan[0] )
print( insan[1:3] )
print( insan[0:4:2] )

insan[0]="rana"
print( insan[0] )

insanlar =[ ["rana","yilmaz"] , ["melek","kara"] , ["zeynep","demir"] ]
print(insanlar)
print(insanlar[0] )
print(insanlar[0][1] )

print( len(insanlar) )
isim="meyra"
print(len(isim))

print( insanlar[-1] )


#hangi indexten başlayacak, kaç indexe kadar(dahil değil) alcak
print(insanlar[-3:-1] )

#hangi indexten başlayacak, hangi indexe kadar alcak, kaçar kaçar atlayacak
print(insanlar[-3:-1:2] )
```

```python


rakamlar=[1,2,3,4,5,6,7,8,9]

print( rakamlar[:-2] )
print( rakamlar[2:] )
print(rakamlar[:])



# Liste Methodları
 
rakamlar.append(64)
print(rakamlar)

rakamlar.remove(64)
print(rakamlar)

rakamlar.pop()
print(rakamlar)

rakamlar.pop(4)
print(rakamlar)

yeni_list=rakamlar.copy()
print(yeni_list)
```

```python

a=[1,2,3,4]
b=[5,6,7,8]
c=a+b
print(c)

a.extend(b)
print(a)

a=[1,2,3,5,5,5]
print( a.count(5))

print(a.index(3))

a.insert(3,"aa")
print(a)

a.pop(3)
print(a)

b=[3,5,2,88,23,67]
b.sort()
print(b)

b.sort(reverse=True)
print(b)
```

```python
isimler=["ali","mehmet","cemil","meyra"]

#diziye göre (index) tersten sıralama
isimler.reverse()
print(isimler)

#alfabetik tersten sıralama 
isimler.sort(reverse=True)
print(isimler)


isimler.clear()
print(isimler)
```
