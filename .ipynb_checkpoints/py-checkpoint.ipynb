##### Encapsulation(Kapsülleme) 

```python
class Calisan:
    def __init__(self,ad,soyad,maas):
        self.ad=ad
        self.soyad=soyad    #global degisken
        self.__maas=maas    #private degisken
        
    def info(self):
        print("{} {} {}".format(self.ad,self.soyad,self.maas))

    #methodları da erişilemez yapmak için  __method 

    def getMaas(self):
        return self.__maas
    def setMaas(self,maas):
        self.__maas=maas
        


calisan=Calisan("ayse","yilmaz",34000)

print(calisan.getMaas() )

calisan.setMaas(31000)
print(calisan.getMaas() )
```

### Kalıtım(Inheritance)

```python
# bir sınıfın (alt sınıf) başka bir sınıfın (üst sınıf) özelliklerini ve metodlarını miras almasını sağlar. Bu, kod tekrarını azaltır ve sınıflar arasında ilişki kurar.

# Alt sınıf, üst sınıfın tüm özelliklerine ve metodlarına erişebilir.
# Alt sınıf, üst sınıfın metodlarını ezebilir (override) veya yeni metodlar ekleyebilir.

# Aynı özellik ve methodları tekrar etmemek için bir ana sınıf oluşturuyoruz
# alt sınıflar ana sınıftan bazı özellik ve methodları miraz alıyor
# super().__init__() ile ana sınıfın özelliklerini alabiliyoruz
# alt sınıfta tekrar tanımlamaya gerek kalmıyor

#class tanımlama
# class turemis(temel):


class Okul:
	def __init__(self,isim,soyisim,yas):
		self.isim=isim
		self.soyisim=soyisim
		self.yas=yas
		print("okul sinifi calisti")
#nesne oluşturma
ahmet =Okul("ahmet","yilmaz",45)
print(ahmet.isim)


class Ogretmen(Okul):
	def __init__(self,isim,soyisim,yas,maas,uzmanlik):
		super().__init__(isim,soyisim,yas)
		#self gerekmiyor		
		self.maas=maas
		self.uzmanlik=uzmanlik

		print("ogretmen sinifi calisti")

	def info(self):
		print("{} {} yasında {}".format(self.isim,self.yas,self.uzmanlik))

semih=Ogretmen("semih","saygin",56,17000,"matematik ogretmeni")
semih.info()
```

### Abstract Class (soyut sınıflar)

```python

# doğrudan nesne oluşturulamayan, yalnızca alt sınıflar için bir şablon görevi gören sınıftır.
# Nesne Üretilemez.Soyut sınıftan doğrudan nesne oluşturulamaz.
# Alt sınıflar bu sınıfı miras alarak soyut metodları uygular.
# Gövdesi olmayan (içeriği tanımlanmamış) metodlar barındırabilir.
```

```python

#modul		#class,method
from abc import ABC,abstractmethod


class Hayvan(ABC):
	#abstract yapılmak istenenin üzerinde belirtliyor

	@abstractmethod
	def yuru(self):
		print("hayvan yuruyor")
	
	#abstract olmayan method da tanımlanabilir 
	#alt sınıflara aktarılır
	def kos(self):
		print("hayvan kosuyor")



class Aslan(Hayvan):
	def yuru(self):
		print("aslan yuruyor")
	
	def kos(self):
		print("aslan kosuyor")


aslan=Aslan()
aslan.yuru()
```

### Overriding(Üzerinde yazma)

```python
class Okul:
	def __init__(self,isim,soyisim,yas):
		self.isim=isim
		self.soyisim=soyisim
		self.yas=yas

		print("okul sinifi calisti")

	def info(self):
		print("{} {} yasında".format(self.isim,self.yas))


class Ogretmen(Okul):
	def __init__(self,isim,soyisim,yas,maas,uzmanlik):
		super().__init__(isim,soyisim,yas)
		#self gerekmiyor		
		
		self.maas=maas
		self.uzmanlik=uzmanlik

		print("ogretmen sinifi calisti")

	def info(self):
		print("{} {} yasında {}".format(self.isim,self.yas,self.uzmanlik))
	#info methodu override edildi

bir =Okul("ali","saglam",34)

iki=Ogretmen("mustafa","yildiz",37,4000,"turkce")
#önce temel sınıf sonra turemis sınıfın constructorı çağrılır

bir.info()
iki.info()
```

### Polymorphism(çok şekillik, biçimlilik)

```python

# farklı sınıflardan nesnelerin aynı isimdeki metodu farklı şekillerde uygulayabilmesidir. Yani, bir metodun farklı sınıflarda farklı işlevler görmesi sağlanır.

# 📌 Türleri:
# Metod Polimorfizmi (Method Overloading): Aynı isimli metodun farklı parametrelerle tanımlanması.
# Metod Geçersiz Kılma (Method Overriding): Üst sınıftaki metodun alt sınıf tarafından yeniden tanımlanması.

```

```python

class Okul:
	def __init__(self,isim,soyisim,yas,maas):
		self.isim=isim
		self.soyisim=soyisim
		self.yas=yas
		self.maas=maas

		print("okul sinifi calisti")

	def info(self):
		print("{} {} yasında ".format(self.isim,self.yas))

	def zam(self):
		return self.maas*1.2


class Ogretmen(Okul):
	def __init__(self,isim,soyisim,yas,maas,uzman):
		super().__init__(isim,soyisim,yas,maas)
		self.uzman=uzman

		print("ogretmen sinifi calisti")

	def info(self):
		print("{} {} yasında {}".format(self.isim,self.yas,self.uzmanlik))

	
	def zam(self):
		return self.maas*1.5


class Yonetim(Okul):
	def __init__(self,isim,soyisim,yas,maas,birim):
		super().__init__(isim,soyisim,yas,maas)
		self.birim=birim

		print("yonetim sinifi calisti")

	def info(self):
		print("{} {} yasında {}".format(self.isim,self.yas,self.birim))

	
	def zam(self):
		return self.maas*2


ahmet=Okul("ahmet","korkmaz",45,5000)
ali=Ogretmen("ali","cesur",25,5000,"matematik")
kemal=Yonetim("kemal","buyuk",56,5000,"müdür")

print(
	ahmet.zam()," ",
	ali.zam()," ",
	kemal.zam()
)
#farklı zam override poliformizm


  
```

### *args and *kwargs

##### Arbitrary Arguments, *args

```python

#bir fonksiyona birden fazla argüman veya anahtar kelime argümanı geçirmeye olanak tanır
#ne kadar parametre verileceği belli olmayan durumlarda

# *args (Arguments - Pozisyonel Argümanlar):
# Bir fonksiyona sınırsız sayıda pozisyonel argüman göndermeyi sağlar.
# Tuple olarak alınır.
```

##### **kwargs (Keyword Arguments - Anahtar Kelime Argümanları):

```python
# Bir fonksiyona sınırsız sayıda anahtar-değer (keyword) argümanı göndermeyi sağlar.
# Dictionary (sözlük) olarak alınır.
```

```python

def topla( *args):
	sum=0
	for i in args:
		sum+=i
	print(sum)
	#tuple

topla(6,7,3,9)

# *args yerine *numaralar da yazılabilir




def isim(*args):
	print("{} {}. ".format(args[0], args[1] ))

isim("meyra", "maden")

```

```python

# **kwargs  -> dictionary şeklinde çıktı

def kimlik( **kwargs):
	print("kwargs: ",kwargs)

kimlik(isim="meyra", soyisim="maden")


def anahtar( **kwargs):
	for key,value in kwargs.items():
		print("anahtar[",key,"] : ",value)

anahtar(isim="meyra",soyisim="maden", yas=20)

```

```python


#sırası önemli
def beraber( *args, **kwargs):
	print("args: ",args , " kwargs: ", kwargs)

beraber(1,2,3,4,5,6,isim="meyra",soyisim="maden")

```

```python


def birlestir(**kwargs):
	son=""
	for kelime in kwargs.values():
		son=son+kelime
	return son

birlestir(a="python ",b="dunyanın ",c="en cok kullanilan ",d="programlama dilidir")
```
