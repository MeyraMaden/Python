### Collection

```python
#sqldeki tablo, nosqldeki collection oluşturma
collection = database["musteri"]

#veritabanındaki collection listesi
clist = database.list_collection_names()
if "musteri" in clist:
  print("collection olusturuldu")
```

### Insert

```python
#document
#tek veri ekleme
data= {"name":"Meyra", "age": 20}       #dictionary
collection.insert_one(data)
#collection.insert_one("name":"Meyra", "age": 20)
```

```python
#birden çok veri ekleme
data2= [
    {"name":"Burak", "age": 20},
    {"name":"Arda", "age": 20},
    {"name":"Ayse"}
]

collection.insert_many(data2)
```

```python
#İD vererek ekleme
data3= {"_id":1, "name":"Yesim", "age": 19}
collection.insert_one(data3)
```

```python
dict={ "_id":2, "name": "fatma" ,"age": 18}
x=collection.insert_one(dict)

#id yazdırma
print (x.inserted_id)
```

### Find

```python
#ilk veriyi döndürme
print(collection.find_one() )
```

```python
#tüm verileri yazdırma
for i in collection.find():
    print(i)
```

```python
#belirli alanlara göre verileri yazdırma

#1.parametre -> sorgu, 
#2.parametre -> sonuca hangi alanların dahil edileceği / boş bırakılırsa tüm alanlar dahil
for i in collection.find({} , {"_id":0 , "name":1}):
    print(i)
# 0-> hariç 1-> sorguya dahil et
#0,1,0 şeklinde yapılamıyor baştaki sıfır zaten hariç anlamına geliyor
```
