### Query

```python
#tekli sorgu
sorgu ={"name":"Meyra"}

for i in collection.find(sorgu, {"_id": 0} ):
    print (i)
```

```python
# $gt  ->greater than 

sorgu= {"age": {"$gt" :19} }
for i in collection.find(sorgu , {"_id": 0} ):
    print (i)
```

```python
# $gte  ->greater than equal  

sorgu= {"age": {"$gte" :19} }
for i in collection.find(sorgu , {"_id": 0} ):
    print (i)
```

```python
# $lt  ->lesser than 

sorgu= {"age": {"$lt" :20} }
for i in collection.find(sorgu, {"_id": 0} ):
    print (i)
```

```python
# $lte  ->lesser than equal

sorgu= {"age": {"$lte" :20} }
for i in collection.find(sorgu , {"_id": 0} ):
    print (i)
```

```python
# $ne  ->not equal

sorgu= {"age": {"$ne" :20} }
for i in collection.find(sorgu , {"_id": 0} ):
    print (i)
```

```python
# $in  ->

sorgu= {"age": {"$in" :[18,19]} }
for i in collection.find(sorgu , {"_id": 0} ):
    print (i)
```

```python
# $nin  ->not in

sorgu= {"age": {"$nin" :[19]} }
for i in collection.find(sorgu , {"_id": 0} ):
    print (i)
```

```python
# düzenli ifadeler

sorgu= {"name": {"$regex" : ".*" } }
for i in collection.find(sorgu , {"_id": 0} ):
    print (i)
```

### Sort

```python
#sıralama 
# descending -> azalan -1
# ascending -> artan 1
# default -> ascending küçükten büyüğe ( 1 )

for i in collection.find().sort("age"):
    print (i)
    
print ("\n\n\n")

for i in collection.find().sort("age",-1):
    print (i)

print ("\n\n\n")

for i in collection.find().sort("age",1):
    print (i)
```

##### and or

```python
# and
for i in collection.find( { "$and": [ {"name":"Meyra"} , {"age":20} ] }):
    print(i)
```

```python
# or
for i in collection.find( { "$or": [ {"name":"Meyra"} , {"age":20} ] }):
    print(i)
```
