##### Tuple(demetler)

```python
#listelerle neredeyse aynı, tek farkı değerler değiştirilemez
#yeni eleman eklenemez
#kullanımda ufak farklılıklar var
#listeler gibi index var, köşeli değil yuvarlak parantezle tanımlanır
#subscriptable (indekslenebilir)

zamirler=("ben","sen","o","biz","siz","onlar")
print( zamirler[2])

#zamirler[0]="kem" 		#hata verir

print( zamirler[0:3])
print( zamirler[-3:-1])

print( len( zamirler) )

sayilar=(1,2,3,4)
birles=zamirler+sayilar
print(birles)

print( sayilar.count(3) )
print ( sayilar.index(4) )

```

##### Kümeler(Sets)

```python
#list e benziyor, {} ile tanımlanır(sözlüklere benzer şekilde)
#aynı elemanları kendisine dahil etmiyor 

sayi={1,3,4,6,6,6,10,84,7,5,56}
print(sayi)

#indexi bulunmuyor,indexlenemez
#print(sayi[1])		#hata

sayi.add(100)
print(sayi)

#elemanı silmek için
sayi.discard(7)
print(sayi)

sayi.remove(4)
print(sayi)

yeni={11,31,41,51,61,101,561,841,1001}
sayi.update(yeni)
print(sayi)
#kümeleri birleştirmiş oluyor
```

```python
#küme işlemleri
a={1,2,4,6}
b={4,5,6,8,9}

#a nın b den farkı
print( a.difference(b) )

#kesişim
print( a.intersection(b) )

#birleşim
print( a.union(b) )

print(len(a))
```
